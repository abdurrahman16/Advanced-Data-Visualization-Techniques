```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dates = pd.date_range(start='2024-01-01', end ='2025-12-31', freq='D')
```

```python
np.random.seed(42)  # for reproducibility

temperature = np.random.uniform(10, 35, len(dates))
sales = np.random.randint(100, 1000, len(dates))
visitors = np.random.randint(500, 5000, len(dates))
revenue = sales * np.random.uniform(5, 50, len(dates))
conversion_rate = np.random.uniform(0.01, 0.1, len(dates))
categories = np.random.choice(['A', 'B', 'C', 'D'], len(dates))

long_time_series_df = pd.DataFrame({
    'Date': dates,
    'Temperature': temperature,
    'Sales': sales,
    'Visitors': visitors,
    'Revenue': revenue,
    'Conversion_Rate': conversion_rate,
    'Category': categories
})

display(long_time_series_df.head())
```

```python
df=long_time_series_df
df.head()
```

```python
df_temp=df[["Date", "Temperature"]]
df_temp.head()
```

```python
df_temp=df_temp.set_index("Date")
df_temp.head()
```

```python
df_temp.plot(figsize=(18,9), color='red', linewidth=1)
plt.title("Temperature Over Time", fontsize=20)
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```
